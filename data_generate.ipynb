```python
import json
import os

# 配置
DATA_PREFIX = '/mnt/workspace/modelscope/data/'
EN_FILE = DATA_PREFIX + "WikiMatrix.en-zh.en"
ZH_FILE = DATA_PREFIX + "WikiMatrix.en-zh.zh"
OUTPUT_DIR = "./test_data"
NUM_TRAIN_SAMPLES = 20000  # 训练使用的样本数（跳过这些）
NUM_TEST_SAMPLES = 1000    # 测试集样本数

os.makedirs(OUTPUT_DIR, exist_ok=True)

def generate_test_dataset():
    """生成测试数据集（JSONL 格式）"""
    print("🔄 Generating test dataset...")
    
    test_data = []
    with open(EN_FILE, encoding="utf-8") as f_en, open(ZH_FILE, encoding="utf-8") as f_zh:
        # 跳过训练集使用的前 NUM_TRAIN_SAMPLES 条
        for i, (en_line, zh_line) in enumerate(zip(f_en, f_zh)):
            if i < NUM_TRAIN_SAMPLES: 
                continue
            
            en_line = en_line.strip()
            zh_line = zh_line.strip()
            
            if en_line and zh_line: 
                test_data.append({
                    'en': en_line,
                    'zh': zh_line
                })
            
            if len(test_data) >= NUM_TEST_SAMPLES:
                break
    
    # 保存为 JSONL
    output_path = f"{OUTPUT_DIR}/en2zh_test.jsonl"
    with open(output_path, 'w', encoding='utf-8') as f:
        for item in test_data: 
            f.write(json.dumps(item, ensure_ascii=False) + '\n')
    
    print(f"✅ Generated {len(test_data)} test samples")
    print(f"✅ Saved to {output_path}")
    
    # 保存示例预览
    print("\n📋 Sample data:")
    for i, item in enumerate(test_data[:3]):
        print(f"\n--- Sample {i+1} ---")
        print(f"EN: {item['en']}")
        print(f"ZH: {item['zh']}")
```

```python
if __name__ == "__main__":
    generate_test_dataset()
```
